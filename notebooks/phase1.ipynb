<br><br>
<b>نوع مطالعه و روش کلی تولید داده:</b> این داده حاصل یک پیمایش و مطالعه مشاهده‌ای مقطعی است (آزمایش تصادفی نیست). داده‌ها از طریق ترکیب مصاحبه‌های جمعیت‌شناختی (DEMO_L)، معاینات فیزیکی و تصویربرداری الاستوگرافی کبد در مراکز معاینه سیار (LUX_L، BMX_L، BPXO_L) و آزمایش‌های خون و بیوشیمی (CBC_L، GHB_L، TCHOL_L، BIOPRO_L) تولید شده‌اند.

<b>محدودیت دسترسی:</b> تمامی فایل‌ها از بخش عمومی وب‌سایت رسمی CDC (Public Data Files) با فرمت استاندارد XPT. دریافت شده‌اند و هیچ‌گونه محدودیت دسترسی وجود ندارد.

<h2>تسک مشترک ۳: نمونه‌گیری، انتخاب و قابلیت تعمیم</h2>
<b>روش انتخاب افراد و نوع نمونه‌گیری:</b> شرکت‌کنندگان در پیمایش NHANES به صورت داوطلبانه یا نمونه‌گیری تصادفی ساده انتخاب نشده‌اند، بلکه از طریق یک طرح نمونه‌گیری احتمالی پیچیده، چندمرحله‌ای و طبقه‌بندی‌شده از جمعیت غیرسازمانی ایالات متحده انتخاب شده‌اند.

## تسک مشترک ۴: طبقه‌بندی متغیرها

##### مشخصات متغیرهای اصلی انتخاب‌شده از ۸ فایل داده به همراه نقش تحلیلی، نوع متغیر و مقیاس اندازه‌گیری آن‌ها در جدول زیر خلاصه شده است:

| نام متغیر | فایل منبع | شرح متغیر | نقش در تحلیل | نوع متغیر | مقیاس اندازه‌گیری |
| :--- | :--- | :--- | :--- | :--- | :--- |
| SEQN | مشترک | شناسه یکتای شرکت‌کننده | کلید اتصال فایل‌ها | شناسه‌ای | اسمی |
| LUXSMED | LUX_L | میانه سفتی کبد (kPa) | متغیر پاسخ اصلی | کمی پیوسته | نسبی |
| NLR | محاسباتی (CBC_L) | نسبت نوتروفیل به لنفوسیت | متغیر توضیحی اصلی | کمی پیوسته | نسبی |
| LBDNENO | CBC_L | شمارش مطلق نوتروفیل | مؤلفه محاسبه NLR | کمی پیوسته | نسبی |
| LBDLYMNO | CBC_L | شمارش مطلق لنفوسیت | مؤلفه محاسبه NLR | کمی پیوسته | نسبی |
| LUXCAPM | LUX_L | میانه شاخص چربی کبد (CAP) | متغیر کنترلی | کمی پیوسته | نسبی |
| RIDAGEYR | DEMO_L | سن به سال کامل | فیلتر جمعیت و کنترل | کمی گسسته | نسبی |
| RIAGENDR | DEMO_L | جنسیت (مرد / زن) | متغیر گروه‌بندی | طبقه‌ای دودویی | اسمی |
| BMXBMI | BMX_L | شاخص توده بدنی (BMI) | متغیر کنترلی | کمی پیوسته | نسبی |
| BPXOSY1 | BPXO_L | فشار خون سیستولیک | متغیر کنترلی | کمی پیوسته | نسبی |
| LBXGH | GHB_L | قند خون سه‌ماهه (HbA1c) | متغیر کنترلی | کمی پیوسته | نسبی |
| LBXTC | TCHOL_L | کلسترول تام خون | متغیر کنترلی | کمی پیوسته | نسبی |
| LBXSATSI | BIOPRO_L | آنزیم کبدی ALT | متغیر کنترلی | کمی پیوسته | نسبی |
| LBXSASSI | BIOPRO_L | آنزیم کبدی AST | متغیر کنترلی | کمی پیوسته | نسبی |
| WTMEC2YR | DEMO_L | وزن نمونه‌گیری دو ساله معاینات | وزن‌دهی برای تعمیم | کمی پیوسته | نسبی |

</div>

```python
import pandas as pd
import numpy as np
from pathlib import Path

# مسیر پوشه داده
data_dir = Path("../data")

# لیست متغیرهای مورد نیاز از هر هشت فایل
files_vars = {
    "DEMO_L.xpt": ["SEQN", "RIDAGEYR", "RIAGENDR", "WTMEC2YR"],
    "LUX_L.xpt": ["SEQN", "LUXSMED", "LUXCAPM"],
    "CBC_L.xpt": ["SEQN", "LBDNENO", "LBDLYMNO"],
    "BMX_L.xpt": ["SEQN", "BMXBMI"],
    "BPXO_L.xpt": ["SEQN", "BPXOSY1"],
    "GHB_L.xpt": ["SEQN", "LBXGH"],
    "TCHOL_L.xpt": ["SEQN", "LBXTC"],
    "BIOPRO_L.xpt": ["SEQN", "LBXSATSI", "LBXSASSI"]
}

# خواندن فایل پایه و اتصال فایلها به یکدیگر
df_merged = None

for filename, cols in files_vars.items():
    file_path = data_dir / filename
    temp_df = pd.read_sas(file_path, format="xport")[cols]
    temp_df["SEQN"] = temp_df["SEQN"].astype(int)
    
    # بررسی یکتا بودن شناسه در هر فایل
    assert temp_df["SEQN"].is_unique, f"Duplicate SEQN found in {filename}"
    
    if df_merged is None:
        df_merged = temp_df
    else:
        df_merged = pd.merge(df_merged, temp_df, on="SEQN", how="left")

# محاسبه شاخص التهابی بر اساس نوتروفیل و لنفوسیت
df_merged["NLR"] = np.where(
    df_merged["LBDLYMNO"] > 0,
    df_merged["LBDNENO"] / df_merged["LBDLYMNO"],
    np.nan
)

print("\u202bتعداد کل افراد در فایل پایه (DEMO_L):\u202c", len(df_merged))
print("تعداد ستون‌ها پس از ادغام:", df_merged.shape[1])
df_merged.head()
```

<h2>تسک مشترک ۵: ممیزی کیفیت داده و تعریف جمعیت تحلیلی (الزام الف-۴)</h2>
در این مرحله، ابتدا یکتا بودن شناسه‌ها (SEQN) و مقادیر ناممکن یا کدهای ویژه بررسی شده، سپس داده‌ها طبق معیارهای ورود و خروج به جامعه هدف پژوهش (بزرگسالان ۲۰ سال و بالاتر) محدود می‌شوند و در نهایت وضعیت داده‌های گمشده (Missing) و نحوه برخورد با آن‌ها گزارش می‌گردد.
</div>

```python
# بررسی رکوردهای تکراری بر اساس شناسه  
duplicate_count = df_merged["SEQN"].duplicated().sum()
print(f"تعداد رکوردهای تکراری : {duplicate_count}")

# فیلتر کردن جامعه هدف: بزرگسالان ۲۰ سال و بالاتر
df_adults = df_merged[df_merged["RIDAGEYR"] >= 20].copy()
print(f"تعداد کل افراد پیش از فیلتر سنی: {len(df_merged)}")
print(f"تعداد بزرگسالان (۲۰ سال و بالاتر): {len(df_adults)}")

# جدول ممیزی داده‌های گمشده در جامعه هدف (بزرگسالان ۲۰ سال و بالاتر)
main_vars = [
    "LUXSMED", "NLR", "LUXCAPM", "RIDAGEYR", "RIAGENDR",
    "BMXBMI", "BPXOSY1", "LBXGH", "LBXTC", "LBXSATSI", "LBXSASSI", "WTMEC2YR"
]

audit_table = pd.DataFrame({
    "تعداد معتبر (Valid N)": df_adults[main_vars].notnull().sum(),
    "تعداد گمشده (Missing N)": df_adults[main_vars].isnull().sum(),
    "درصد گمشده (Missing %)": (df_adults[main_vars].isnull().mean() * 100).round(2),
    "کمینه (Min)": df_adults[main_vars].min().round(2),
    "بیشینه (Max)": df_adults[main_vars].max().round(2)
})

audit_table
```

```python
# گزارش مراحل پالایش و تعداد افراد حذف‌شده (الزام الف-۴)
n_total = len(df_merged)
n_adults = len(df_adults)
print(f"\u202b۱. تعداد کل افراد پس از ادغام فایل‌ها: {n_total} نفر\u202c")
print(f"\u202b۲. تعداد افراد حذف‌شده زیر ۲۰ سال: {n_total - n_adults} نفر (باقی‌مانده بزرگسالان: {n_adults} نفر)\u202c")

# بررسی افراد فاقد وزن معاینه پزشکی (WTMEC2YR == 0)
no_mec_count = (df_adults["WTMEC2YR"] == 0).sum()
df_mec = df_adults[df_adults["WTMEC2YR"] > 0].copy()
print(f"\u202b۳. تعداد بزرگسالان فاقد معاینه در مرکز سیار (WTMEC2YR = 0): {no_mec_count} نفر (باقی‌مانده: {len(df_mec)} نفر)\u202c")

# حذف مقادیر گمشده در دو متغیر اصلی پژوهش (LUXSMED و NLR)
df_clean = df_mec.dropna(subset=["LUXSMED", "NLR"]).copy()
dropped_main = len(df_mec) - len(df_clean)
print(f"\u202b۴. تعداد افراد حذف‌شده به دلیل گمشدگی در متغیرهای اصلی (LUXSMED یا NLR): {dropped_main} نفر\u202c")
print(f"\u202b۵. حجم نمونه تحلیلی نهایی (df_clean): {len(df_clean)} نفر\u202c")
```

<div dir="rtl" align="right">
<b>۱. بررسی رکوردهای تکراری و ناسازگاری‌های منطقی:</b> بررسی ستون شناسه (SEQN) نشان داد هیچ رکورد تکراری در داده وجود ندارد. همچنین پس از اعمال فیلتر سنی، تمامی متغیرها در بازه منطقی قرار دارند (کمینه سن ۲۰ و بیشینه ۸۰ سال است).
<br><br>
<b>۲. بررسی مقادیر غیرممکن و کدهای ویژه:</b> در فایل‌های آزمایشگاهی و الاستوگرافی NHANES، مقادیر نامعتبر به صورت NaN ثبت شده‌اند و کدهای عددی ویژه (مانند 777 یا 999) در متغیرهای کمی منتخب مشاهده نشد. مقادیر کمینه و بیشینه (مانند سفتی کبد از ۱.۵ تا ۷۵ کیلوپاسکال و NLR از ۰.۳ تا ۳۳.۵) از نظر زیستی و بالینی ممکن هستند، اما نشان‌دهنده چولگی به راست و وجود مقادیر حدی در توزیع می‌باشند.
<br><br>
<b>۳. تصمیم آماری برای برخورد با داده‌های گمشده و دلیل آن:</b>
<br>
• <b>متغیرهای اصلی (LUXSMED و NLR):</b> از آنجا که هدف اصلی پژوهش بررسی رابطه التهاب سیستمیک (NLR) و سفتی کبد (LUXSMED) است، جایگذاری مقادیر گمشده در این دو متغیر اصلی باعث خطا در بررسی رابطه واقعی بین این دو متغیر می شود؛ بنابراین از روش تحلیل موارد کامل استفاده شد و ۲۴۴۹ نفری که در یکی از این دو متغیر مقدار گمشده داشتند حذف شدند تا حجم نمونه تحلیلی نهایی به ۵۳۶۰ نفر برسد.
<br>
• <b>متغیرهای کمکی و کنترلی:</b> برای جلوگیری از کاهش غیرضروری حجم نمونه در تحلیل‌های اصلی، ردیف‌هایی که صرفاً در متغیرهای فرعی (مانند کلسترول یا آنزیم‌های کبدی) دارای مقدار گمشده هستند در مجموعه داده نهایی (df_clean) حفظ شدند و تنها در تحلیل‌های چندمتغیره مربوط به همان متغیرها کنار گذاشته می‌شوند.

<h2>تسک مشترک ۶: جدول‌های فراوانی و توزیع تجربی</h2>
در این مرحله، توزیع تجربی متغیرهای پژوهش در نمونه تحلیلی نهایی (۵۳۶۰ نفر) بررسی می‌شود. ابتدا برای متغیرهای طبقه‌ای (جنسیت و طبقه‌بندی بالینی سفتی کبد) جدول فراوانی و درصد ارائه شده و سپس برای متغیرهای کمی، خلاصه توزیع تجربی شامل چندک‌ها (صدک‌های ۱۰، ۲۵، ۵۰، ۷۵ و ۹۰) و ضریب چولگی گزارش می‌گردد.
</div>

```python
# ۱. جدول فراوانی و درصد برای متغیر طبقه‌ای جنسیت (RIAGENDR)
gender_map = {1.0: "مرد (Male)", 2.0: "زن (Female)"}
gender_series = df_clean["RIAGENDR"].map(gender_map)

freq_gender = pd.DataFrame({
    "فراوانی (تعداد)": gender_series.value_counts(),
    "درصد (%)": (gender_series.value_counts(normalize=True) * 100).round(2)
})
print("--- جدول فراوانی و درصد جنسیت در نمونه تحلیلی نهایی ---")
display(freq_gender)

# ۲. جدول فراوانی و درصد برای طبقه‌بندی بالینی سفتی کبد (LUXSMED)
fibrosis_bins = [0, 7.0, 8.2, np.inf]
fibrosis_labels = ["طبیعی (< 7.0 kPa)", "ملایم تا متوسط (7.0 - 8.2 kPa)", "فیبروز پیشرفته (>= 8.2 kPa)"]
fibrosis_cat = pd.cut(df_clean["LUXSMED"], bins=fibrosis_bins, labels=fibrosis_labels, right=False)

freq_fibrosis = pd.DataFrame({
    "فراوانی (تعداد)": fibrosis_cat.value_counts(),
    "درصد (%)": (fibrosis_cat.value_counts(normalize=True) * 100).round(2)
})
print("\n--- جدول فراوانی وضعیت سفتی کبد (بر اساس آستانه‌های بالینی) ---")
display(freq_fibrosis)

# ۳. خلاصه توزیع تجربی (چندک‌ها و ضریب چولگی) برای متغیرهای کمی
quant_vars = [
    "LUXSMED", "NLR", "LUXCAPM", "RIDAGEYR", 
    "BMXBMI", "BPXOSY1", "LBXGH", "LBXTC", "LBXSATSI", "LBXSASSI"
]

dist_summary = pd.DataFrame({
    "تعداد معتبر (N)": df_clean[quant_vars].notnull().sum(),
    "صدک ۱۰ (P10)": df_clean[quant_vars].quantile(0.10).round(2),
    "چارک اول (Q1)": df_clean[quant_vars].quantile(0.25).round(2),
    "میانه (Median)": df_clean[quant_vars].quantile(0.50).round(2),
    "چارک سوم (Q3)": df_clean[quant_vars].quantile(0.75).round(2),
    "صدک ۹۰ (P90)": df_clean[quant_vars].quantile(0.90).round(2),
    "ضریب چولگی (Skewness)": df_clean[quant_vars].skew().round(2)
})

print("\n--- خلاصه توزیع تجربی متغیرهای کمی ---")
display(dist_summary)
```

<div dir="rtl" align="right">
<b>الگوهای مهم در متغیرهای طبقه‌ای:</b>
<br>
• <b>توازن جنسیتی در نمونه تحلیلی:</b> بررسی جدول فراوانی جنسیت نشان می‌دهد که ترکیب نمونه تحلیلی نهایی میان زنان و مردان متعادل است (با فراوانی کمی بیشتر برای زنان)، که امکان مقایسه توصیفی معتبر بین دو جنس را در مراحل بعد فراهم می‌کند.
<br>
• <b>توزیع سطوح سفتی کبد:</b> بخش عمده‌ای از بزرگسالان شرکت‌کننده در بازه طبیعی سفتی کبد (کمتر از ۷ کیلوپاسکال) قرار دارند و تنها بخش کوچکی از جمعیت در رده فیبروز پیشرفته (۸.۲ کیلوپاسکال و بالاتر) جای می‌گیرند که با الگوی شیوع بیماری‌های کبدی در جمعیت عمومی هم‌خوانی دارد.
<br><br>
<b>الگوهای مهم در توزیع تجربی متغیرهای کمی:</b>
<br>
• <b>چولگی شدید به راست در متغیرهای اصلی و آنزیم‌های کبدی:</b> بررسی چندک‌ها و ضریب چولگی نشان می‌دهد متغیرهای سفتی کبد (LUXSMED)، نسبت نوتروفیل به لنفوسیت (NLR) و آنزیم‌های کبدی (LBXSATSI و LBXSASSI) دارای چولگی مثبت (به سمت راست) هستند؛ به طوری که فاصله صدک ۹۰ تا میانه به مراتب بزرگ‌تر از فاصله میانه تا صدک ۱۰ است. این الگو نشان می‌دهد اکثر افراد جامعه مقادیر پایین و نرمال دارند، اما وجود تعداد کمی از افراد با التهاب یا سفتی کبد بسیار بالا، دنباله توزیع را به سمت راست کشیده است.
<br>
• <b>توزیع متقارن‌تر در متغیرهای عمومی:</b> متغیرهایی مانند شاخص چربی کبد (LUXCAPM)، کلسترول کل (LBXTC) و فشار خون سیستولیک (BPXOSY1) ضریب چولگی نزدیک به صفر یا ملایمی دارند و چارک‌های اول و سوم آن‌ها با فاصله نسبتاً متوازنی در دو طرف میانه قرار گرفته‌اند.

<h2>تسک مشترک ۷: مرکز و پراکندگی و مقایسه خلاصه خام و وزن‌دار (الزام الف-۳)</h2>
ابتدا شاخص‌های مرکزی (میانگین و میانه) و شاخص‌های پراکندگی (انحراف معیار، چارک‌ها و دامنه بین‌چارکی) برای متغیرهای کمی محاسبه می‌شوند تا معیار مناسب برای هر متغیر بر اساس شکل توزیع آن مشخص گردد. سپس در راستای الزام ویژه مسیر الف (NHANES)، مقادیر میانگین ساده (بدون وزن) و میانگین وزن‌دار (با اعمال وزن معاینه پزشکی WTMEC2YR) در کنار هم گزارش و مقایسه می‌شوند.
</div>

```python
# ۱. محاسبه شاخص‌های مرکز و پراکندگی برای متغیرهای کمی
q1 = df_clean[quant_vars].quantile(0.25)
q3 = df_clean[quant_vars].quantile(0.75)

center_disp_table = pd.DataFrame({
    "میانگین (Mean)": df_clean[quant_vars].mean().round(2),
    "انحراف معیار (SD)": df_clean[quant_vars].std().round(2),
    "میانه (Median)": df_clean[quant_vars].median().round(2),
    "چارک اول (Q1)": q1.round(2),
    "چارک سوم (Q3)": q3.round(2),
    "دامنه بین‌چارکی (IQR)": (q3 - q1).round(2)
})

print("--- جدول شاخص‌های مرکز و پراکندگی متغیرهای کمی ---")
display(center_disp_table)

# ۲. مقایسه خلاصه خام (بدون وزن) و وزن‌دار (الزام الف-۳)
def weighted_mean(df, col, weight_col="WTMEC2YR"):
    valid = df[[col, weight_col]].dropna()
    return np.average(valid[col], weights=valid[weight_col])

compare_vars = ["LUXSMED", "NLR", "LUXCAPM", "RIDAGEYR", "BMXBMI", "BPXOSY1"]

weighted_comp_table = pd.DataFrame({
    "میانگین خام (Unweighted Mean)": [df_clean[c].mean() for c in compare_vars],
    "میانگین وزن‌دار (Weighted Mean)": [weighted_mean(df_clean, c) for c in compare_vars]
}, index=compare_vars).round(2)

weighted_comp_table["تفاوت (وزن‌دار - خام)"] = (
    weighted_comp_table["میانگین وزن‌دار (Weighted Mean)"] - 
    weighted_comp_table["میانگین خام (Unweighted Mean)"]
).round(2)

# محاسبه نسبت خام و وزن‌دار برای فیبروز پیشرفته (>= 8.2 kPa)
df_clean["adv_fibrosis"] = (df_clean["LUXSMED"] >= 8.2).astype(int)
unw_prop = df_clean["adv_fibrosis"].mean() * 100
w_prop = weighted_mean(df_clean, "adv_fibrosis") * 100

print("\n--- جدول مقایسه میانگین خام و وزن‌دار (با اعمال وزن WTMEC2YR) ---")
display(weighted_comp_table)
print(f"\u202bدرصد فیبروز پیشرفته (>= 8.2 kPa) - خام: {unw_prop:.2f}% | وزن‌دار: {w_prop:.2f}%\u202c")
```

• <b>متغیرهای دارای چولگی شدید (مانند LUXSMED، NLR و آنزیم‌های کبدی):</b> در این متغیرها، میانگین به طور محسوسی بزرگ‌تر از میانه است و انحراف معیار نیز تحت تأثیر مقادیر حدی (پرت) در دنباله راست توزیع، بزرگ شده است. از آنجا که میانگین و انحراف معیار نسبت به مشاهدات پرت بسیار حساس هستند، برای توصیف مرکز و پراکندگی این متغیرها، <b>میانه (Median)</b> و <b>دامنه بین‌چارکی (IQR)</b> معیارهای مقاوم‌تر و مناسب‌تری محسوب می‌شوند.

• <b>متغیرهای دارای توزیع نزدیک به متقارن (مانند LUXCAPM، کلسترول و فشار خون):</b> در این متغیرها، مقادیر میانگین و میانه به یکدیگر بسیار نزدیک هستند و چولگی شدیدی وجود ندارد؛ بنابراین گزارش <b>میانگین (Mean)</b> و <b>انحراف معیار (SD)</b> در کنار میانه و دامنه بین چارکی  کاملاً گویا و مناسب است.

<b>۲. مقایسه خلاصه خام و وزن‌دار (الزام الف-۳ و نقش طرح نمونه‌گیری):</b>
<br>
• در پیمایش NHANES، به دلیل بیش‌نمونه‌گیری هدفمند از برخی زیرگروه‌ها (مانند سالمندان یا اقلیت‌های نژادی)، ترکیب افراد حاضر در فایل داده با ترکیب واقعی جمعیت ایالات متحده یکسان نیست. برای مثال، میانگین سنی در نمونه خام بالاتر از میانگین سنی وزن‌دار است، زیرا افراد مسن‌تر با احتمال بیشتری وارد نمونه شده‌اند.

• با اعمال وزن معاینه پزشکی (WTMEC2YR)، اثر این بیش‌نمونه‌گیری تعدیل شده و برآوردها (مانند میانگین سن، سفتی کبد، NLR و درصد شیوع فیبروز پیشرفته) به مقادیر واقعی جامعه بزرگسالان ۲۰ سال و بالاتر در ایالات متحده نزدیک می‌شوند.
